# 정적 크롤링을 이용한 웹 데이터 수집

## 상세 페이지에서 쪽수 정보 가져오기

### 1단계: 웹 페이지에서 html 가져오기

```python
# 필요한 라이브러리 임포트
import requests
```

```python
# 물고기는 존재하지 않는다.의 상세페이지 정보 수집하기
url = "https://www.yes24.com/product/goods/105526047"

## requests.get() 함수 호출 -> html획득
html = requests.get(url).text

## 결과 확인
print(f'추출한 상세 페이지 html: \n{html}')
```

### 2단계: BeautifulSoup 객체 생성

```python
# 필요한 클래스 임포트
from bs4 import BeautifulSoup

# 클래스 호출, 객체 생성, 분석결과 저장
soup = BeautifulSoup(html, 'html.parser')
```

### 3단계: 원하는 데이터 추출하기

```python
# 쪽수 정보가 들어있는 태그 정보 추출 -> select('CSS 선택자') -> 결과값: 리스트(여러 값)
# select 함수는 여러개가 나올 수 있다.
tags = soup.select('td.txt.lastCol')
print(f'추출한 tag 정보확인: \n{tags}')
```

```python
# 쪽수 정보가 들어있는 태그 -> 인덱싱 -> 두 번째 성분원소 추출
target = tags[1]
print(target)
```

```python
# 쪽수 정보가 들어있는 태그 -> 텍스트 추출 -> .text 또는 .get_text()
text = target.get_text()
print(f'태그로부터 추출한 텍스트 데이터: \n{text}')
```

```python
# 쪽수 정보 추출

## 방법1 -> 슬라이싱 -> start:stop-1
page1=text[0:4]
print(f'쪽수 정보: {page1}')

print('-'*80)

## 방법2 -> 스플릿 -> slint()
page2 = text.split(' | ')[0]
print(f'쪽수 정보: {page2}')
```
